Revisamos la cantidad de valores únicos en las variables más relevantes, se excluyen las de geolocalización o ubicación.

```python
tienda.nunique()
```

```python
columns_to_check = ["Producto", "Categoría del Producto", "Precio", "Costo de envío",
                    "Fecha de Compra", "Vendedor", "ciudad", "Calificación",
                    "Método de pago", "Cantidad de cuotas"]

print("Cantidad de datos únicos por columna:")
for col in columns_to_check:
    unique_count = tienda[col].nunique()
    print(f"- {col}: {unique_count}")

print("\nColumnas con menos de 20 datos únicos y sus valores:")
for col in columns_to_check:
    unique_count = tienda[col].nunique()
    if unique_count < 20:
        unique_values = tienda[col].unique()
        print(f"- {col} ({unique_count} únicos): {unique_values}")
```

```python
# Colectar los valores de las columnas y verificar
columnas = list(tienda.columns)
columnas

# Destrincar las listas con explode
datos = tienda.explode(columnas[1:])
# Resetear los index de las líneas
datos.reset_index(drop=True,inplace=True)
# Observar el DataFrame
datos
```

En la BD tienda no veo necesidad de transformar ni manipular los datos.

```python
import seaborn as sns
import matplotlib.pyplot as plt

avg_sell = tienda.groupby('ciudad')['Precio'].mean().reset_index()
sns.barplot(x='ciudad',y='Precio',data=avg_sell)
plt.title('Promedio del precio de venta de cada producto en las ciudades de la tienda')
plt.xlabel('Ciudad')
plt.xticks(rotation=70)
plt.ylabel('Precio promedio de venta de productos')
plt.show()

sns.countplot(data=tienda, x='ciudad')
plt.title('Ventas de productos por ciudad')
plt.xlabel('Ciudad')
plt.xticks(rotation=70)
plt.ylabel('Ventas')
plt.show()
```

```python
# Calcular el promedio de Costo de envío por ciudad
promedio_costo_envio_por_ciudad = tienda.groupby('ciudad')['Costo de envío'].mean().reset_index()

# Gráfico del promedio de Costo de envío por ciudad
plt.figure(figsize=(12, 6))
sns.barplot(data=promedio_costo_envio_por_ciudad, x='ciudad', y='Costo de envío', palette='viridis')
plt.title('Promedio de Costo de Envío por Ciudad')
plt.xlabel('Ciudad')
plt.ylabel('Costo de Envío Promedio')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```
